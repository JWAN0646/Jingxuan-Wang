```python
import geopandas as gpd

path = "/Users/wangjingxuan/Jingxuan Wang/Surface_Salinity_current_DPIRD_039_WA_GDA2020_Public_Geopackage/Surface_Salinity_current_DPIRD_039_WA_GDA2020_Public.gpkg"

gdf = gpd.read_file(path)

gdf.head()
```

```python
gdf.plot(
    column="surf_sa_co",
    legend=True,
    figsize=(10,8)
)
```

```python
print(gdf.crs)
```

```python
gdf = gdf.to_crs(3857)

gdf["area_km2"] = gdf.geometry.area / 1e6

area_summary = (
    gdf.groupby("surf_sa_co")["area_km2"]
       .sum()
       .reset_index()
       .sort_values("area_km2", ascending=False)
)

area_summary
```

```python
import matplotlib.pyplot as plt

area_summary.plot(
    x="surf_sa_co",
    y="area_km2",
    kind="bar"
)

plt.title("WA Surface Salinity Area Distribution")
plt.show()
```
